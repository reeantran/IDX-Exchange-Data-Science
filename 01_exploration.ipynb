```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df1 = pd.read_csv("CRMLSSold202402.csv")
df2 = pd.read_csv("CRMLSSold202408.csv")
df3 = pd.read_csv("CRMLSSold202409.csv")
df4 = pd.read_csv("CRMLSSold202410.csv")
df5 = pd.read_csv("CRMLSSold202411.csv")
df6 = pd.read_csv("CRMLSSold202412.csv")

merged_df1 = pd.concat([df1, df2], ignore_index=True)
merged_df2 = pd.concat([df3, df4], ignore_index=True)
merged_df3 = pd.concat([df5, df6], ignore_index=True)

merged_df_1 = pd.concat([merged_df1, merged_df2], ignore_index=True)
df = pd.concat([merged_df_1, merged_df3], ignore_index=True)
```

```python
restrict_df = df[
    (df['PropertyType'] == 'Residential') &
    (df['PropertySubType'] == 'SingleFamilyResidence') &
    (df['ClosePrice'] < 20_000_000) &
    (df['LivingArea'] < 20_000) &
    (df['BathroomsTotalInteger'] < 25) &
    (df['LotSizeSquareFeet'] < 750_000_000)
]

closePrice_df = restrict_df[['ClosePrice']]
livingArea_df = restrict_df[['LivingArea']]
bedrooms_df = restrict_df[['BedroomsTotal']]
bathrooms_df = restrict_df[['BathroomsTotalInteger']]
lotSize_df = restrict_df[['LotSizeSquareFeet']]

%store restrict_df 
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(livingArea_df, closePrice_df)
plt.xlabel("Living Area")
plt.ylabel("Close Price")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(bedrooms_df, closePrice_df)
plt.xlabel("Total Bedrooms")
plt.ylabel("Close Price")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(bathrooms_df, closePrice_df)
plt.xlabel("Total Bathrooms")
plt.ylabel("Close Price")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(lotSize_df, closePrice_df)
plt.xlabel("Lot Size")
plt.ylabel("Close Price")
plt.show()
```
