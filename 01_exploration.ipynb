```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("CRMLSSold202401_filled.csv")
print(df.head)
```

```python
restrict_df = df[(df['PropertyType'] == 'Residential') & (df['PropertySubType'] == 'SingleFamilyResidence')]

closePrice_df = restrict_df['ClosePrice']
livingArea_df = restrict_df['LivingArea']
bedrooms_df = restrict_df['BedroomsTotal']
bathrooms_df = restrict_df['BathroomsTotalInteger']
lotSize_df = restrict_df['LotSizeAcres']
```

```python
plt.scatter(livingArea_df, closePrice_df)
plt.xlabel("Living Area")
plt.ylabel("Close Price")
plt.show()
```

```python
plt.scatter(bedrooms_df, closePrice_df)
plt.xlabel("Bedrooms")
plt.ylabel("Close Price")
plt.show()
```

```python
plt.scatter(bathrooms_df, closePrice_df)
plt.xlabel("Bathrooms")
plt.ylabel("Close Price")
plt.show()
```

```python
plt.scatter(lotSize_df, closePrice_df)
plt.xlabel("Lot Size")
plt.ylabel("Close Price")
plt.show()
```
